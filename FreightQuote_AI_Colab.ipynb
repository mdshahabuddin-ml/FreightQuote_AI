## 2. Get the project folder
Upload `FreightQuote_AI.zip` when asked (skipped if the folder already exists). Your existing `freightquote.db` is kept.

```python
import zipfile, shutil
from pathlib import Path

PROJECT = Path("/content/FreightQuote_AI")

if not (PROJECT / "run_colab.py").exists():
    from google.colab import files
    uploaded = files.upload()  # choose FreightQuote_AI.zip
    zip_name = next(iter(uploaded))
    with zipfile.ZipFile(zip_name) as z:
        for member in z.namelist():
            target = Path("/content") / member
            # never overwrite an existing user database
            if target.name == "freightquote.db" and target.exists():
                continue
            z.extract(member, "/content")

print("Project files:")
for p in sorted(PROJECT.rglob("*")):
    if p.is_file() and ".gitkeep" not in p.name:
        print(" ", p.relative_to(PROJECT))
```
